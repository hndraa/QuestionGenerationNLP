```python
import pandas as pd
import numpy as np

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

# SEBELUM VALIDASI GURU

```python
# =========================================
# LOAD DATA
# =========================================

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/H3_rulebased.csv")

print("Total data:", len(df))
```

```python
# =========================================
# LOAD MODEL EMBEDDING
# =========================================

model = SentenceTransformer(
    'sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2'
)
```

```python
# =========================================
# HITUNG COSINE SIMILARITY
# ANTARA KALIMAT ASAL
# DAN PERTANYAAN
# =========================================

scores = []

for i in range(len(df)):

    # =====================================
    # AMBIL KALIMAT ASAL
    # =====================================

    kalimat_asal = str(
        df.iloc[i]["kalimat"]
    )

    # =====================================
    # AMBIL PERTANYAAN
    # =====================================

    pertanyaan = str(
        df.iloc[i]["pertanyaan"]
    )

    # =====================================
    # EMBEDDING
    # =====================================

    kalimat_embedding = model.encode(
        [kalimat_asal]
    )

    pertanyaan_embedding = model.encode(
        [pertanyaan]
    )

    # =====================================
    # COSINE SIMILARITY
    # =====================================

    similarity = cosine_similarity(
        kalimat_embedding,
        pertanyaan_embedding
    )[0][0]

    scores.append(
        round(float(similarity), 4)
    )
```

```python
# =========================================
# SIMPAN SCORE
# =========================================

df["cosine_similarity"] = scores
```

```python
# =========================================
# OUTPUT
# =========================================

print("=" * 50)
print("HASIL COSINE SIMILARITY")
print("=" * 50)

print(df[
    [
        "kalimat",
        "pertanyaan",
        "cosine_similarity"
    ]
].head(20))

print("\nRata-rata cosine similarity:")
print(
    round(
        df["cosine_similarity"].mean(),
        4
    )
)
```

```python
# =========================================
# SIMPAN FILE
# =========================================

df.to_csv(
    "/content/drive/MyDrive/Colab Notebooks/H8_cosine_similarity.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nFile berhasil disimpan!")
```

# SETELAH VALIDASI GURU

```python
# =========================================
# LOAD HASIL COSINE SIMILARITY
# =========================================

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/H4_cosine_similarity.csv")

print("Total data:", len(df))
```

```python
# =========================================
# TENTUKAN THRESHOLD
# =========================================

threshold = 0.34

# =========================================
# TENTUKAN VALID / TIDAK VALID
# =========================================

df["valid"] = df["cosine_similarity"].apply(

    lambda x:
    "Valid"
    if x >= threshold
    else "Tidak Valid"
)
```

```python
# =========================================
# HITUNG JUMLAH
# PERTANYAAN VALID
# =========================================

Qb = len(
    df[
        df["cosine_similarity"] >= threshold
    ]
)
```

```python
# =========================================
# HITUNG TOTAL PERTANYAAN
# =========================================

Qt = len(df)
Q = (Qb / Qt) * 100
```

```python
# =========================================
# OUTPUT
# =========================================

print("=" * 50)
print("EVALUASI PERTANYAAN")
print("=" * 50)

print(f"Threshold                : {threshold}")

print(f"Jumlah Total Pertanyaan  : {Qt}")

print(f"Jumlah Pertanyaan Benar  : {Qb}")

print(f"Jumlah Pertanyaan Salah  : {Qt - Qb}")

print(f"Persentase Benar         : {Q:.2f}%")

print("\nRata-rata score naturalitas:")
print(df["cosine_similarity"].mean())
```

```python
# =========================================
# SIMPAN HASIL
# =========================================

df.to_csv(
    "/content/drive/MyDrive/Colab Notebooks/H4_Dataset_Cosine.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nFile evaluasi berhasil disimpan!")
```
