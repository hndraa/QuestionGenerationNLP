```python
import pandas as pd
import stanza
```

```python
df = pd.read_csv("1.bakal_datasetRB.csv")

print("Total data awal:", len(df))
```

```python
def is_valid(kalimat):
    return isinstance(kalimat, str) and len(kalimat.split()) >= 3

df = df[df["kalimat"].apply(is_valid)].copy()

print("Setelah filter:", len(df))
```

```python
nlp = stanza.Pipeline(
    'id',
    processors='tokenize,pos,lemma,depparse',
    use_gpu=True
)
```

```python
# nlp = stanza.Pipeline(
#     lang='id',
#     dir='C:/stanza_resources',
#     processors='tokenize,pos,lemma,depparse',
#     use_gpu=True
# )
```

```python
hasil = []

for idx, row in df.iterrows():
    kalimat = row["kalimat"]

    try:
        doc = nlp(kalimat)

        for sentence in doc.sentences:
            for word in sentence.words:
                hasil.append({
                    "id_teks": row["id_teks"],
                    "id_kalimat": row["id_kalimat"],
                    "kalimat": kalimat,
                    "kata": word.text,
                    "lemma": word.lemma,
                    "pos": word.upos,
                    "head": word.head,
                    "deprel": word.deprel
                })

    except Exception as e:
        print(f"Error di kalimat: {kalimat}")
        print(e)
```

```python
df_hasil = pd.DataFrame(hasil)

df_hasil["id"] = df_hasil.groupby(["id_teks", "id_kalimat"]).cumcount() + 1

df_hasil.to_csv("hasil_stanza.csv", index=False)

print("Selesai! File: hasil_stanza.csv")
```
