```python
import pandas as pd
```

```python
# =========================
# LOAD DATASET COSINE
# =========================

df = pd.read_csv("H4_Dataset_Cosine.csv")

print("=" * 50)
print("TOTAL DATA AWAL")
print("=" * 50)

print(len(df))
```

```python
# =========================
# CEK KOLOM VALID
# =========================

if "valid" not in df.columns:
    raise ValueError(
        "Kolom 'valid' tidak ditemukan!"
    )

# =========================
# BERSIHKAN NILAI VALID
# =========================

df["valid"] = (
    df["valid"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# =========================
# FILTER DATA VALID SAJA
# =========================

df = df[
    df["valid"] == "valid"
]

print("\nJumlah data valid:")
print(len(df))

# =========================
# HAPUS DATA KOSONG
# =========================

df = df.dropna(
    subset=["kalimat", "pertanyaan"]
)

# =========================
# BERSIHKAN SPASI
# =========================

df["kalimat"] = (
    df["kalimat"]
    .astype(str)
    .str.strip()
)

df["pertanyaan"] = (
    df["pertanyaan"]
    .astype(str)
    .str.strip()
)

# =========================
# HAPUS DUPLIKAT
# =========================

df = df.drop_duplicates(
    subset=["kalimat", "pertanyaan"]
)

# =========================
# DETEKSI TIPE PERTANYAAN
# =========================

def get_tipe_pertanyaan(q):

    q = str(q).lower().strip()

    if q.startswith("siapa"):
        return "siapa"

    elif q.startswith("apa"):
        return "apa"

    elif q.startswith("di mana"):
        return "dimana"

    elif q.startswith("ke mana"):
        return "kemana"

    elif q.startswith("dari mana"):
        return "darimana"

    elif q.startswith("kapan"):
        return "kapan"

    else:
        return "lainnya"

# =========================
# BUAT KOLOM TIPE
# =========================

df["tipe"] = df["pertanyaan"].apply(
    get_tipe_pertanyaan
)

# =========================
# HAPUS TIPE LAINNYA
# =========================

df = df[
    df["tipe"] != "lainnya"
]

# =========================
# FORMAT INPUT MODEL
# =========================

df["input"] = (
    "generate "
    + df["tipe"]
    + ": "
    + df["kalimat"]
)

# =========================
# TARGET
# =========================

df["target"] = df["pertanyaan"]

# =========================
# AMBIL FINAL DATASET
# =========================

dataset_ml = df[
    ["input", "target"]
]

# =========================
# HAPUS DUPLIKAT FINAL
# =========================

dataset_ml = dataset_ml.drop_duplicates()

dataset_ml = dataset_ml.reset_index(
    drop=True
)
```

```python
# =========================
# SIMPAN CSV
# =========================

dataset_ml.to_csv(
    "H5_dataset_ML.csv",
    index=False,
    encoding="utf-8-sig"
)

# =========================
# OUTPUT
# =========================

print("\n")
print("=" * 50)
print("HASIL DATASET ML")
print("=" * 50)

print(dataset_ml.head(10))

print("\nJumlah data akhir:")
print(len(dataset_ml))
```
